The data covers all substates/regions across the USA and provides information on the number of participants and households receiving SNAP benefits.
```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df_snap = pd.read_excel('SNAP - National Data Bank JUL 2023.xlsx', header=2, skiprows=[0])
df_snap.head()
```

```python
df_snap = df_snap.dropna(axis=1, how='all')
df_snap.head()
```

```python
df_snap = df_snap[df_snap['Substate/Region'].astype(str).str.startswith('13')]
df_snap.head(n=10)
```

```python
df_snap = df_snap.drop(columns=['Substate/Region'])
df_snap.head()
```

Although the dataset does not contain any missing data, it needed to be cleaned for only the relevant data. The first change I made was to remove all rows that did not contain information regarding Georgia. The Second change was to drop the empty columns. Although keeping all region-identifying data in a single column might have been beneficial, it was too clustered and contained unnecessary information. As a result, I took the relevant information (identification code and County) and stored them in separate columns. Finally, I deleted the old column (Substate/Region).

```python
import pandas as pd
df = pd.read_csv('Head count of Free and Reduced Lunch.csv')
df.head()
```

##### Clustering - K-means

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

# Select relevant numerical features for clustering
features = ['Percentage (%) of Free & Reduced', 'PK-12 Students Count']
X = df[features].copy()

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Determine optimal number of clusters using the Elbow Method
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()

# Based on the elbow plot k=3 is picked
k = 3
kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

df['KMeans_Cluster'] = clusters

# Visualize the K-Means clusters
plt.figure(figsize=(12, 8))
sns.scatterplot(x='Percentage (%) of Free & Reduced', y='PK-12 Students Count', hue='KMeans_Cluster', data=df, palette='viridis', s=100, alpha=0.8)
plt.title(f'K-Means Clustering with k={k}')
plt.xlabel('Percentage (%) of Free & Reduced')
plt.ylabel('PK-12 Students Count')
plt.grid(True)
plt.show()

print(df['KMeans_Cluster'].value_counts())
```

##### Clustering Findings

Based on the k-means graph, the hypothesis does not seem to be strongly supported. Cluster 0 has a very high percentage of free & reduced lunch; however, their student count varies. This contradicts the idea that schools with higher populations will have higher percentages. Cluster 1 has a lower percentage of free & reduced lunch, but it also covers a broad spectrum. Finally, cluster 2 spans from a low to a high percentage of free & reduced lunch; however, these schools generally have higher student counts than the other clusters. This observation directly contradicts the hypothesis.

This clustering suggests that the percentage of free & reduced lunch is not directly or positively correlated with the student count, and that any slight correlation indicates that smaller schools tend to have a higher free & reduced lunch percentage.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

features = [
    "Child Food Insecurity Rate",
    "% food insecure children in HH w/ HH incomes below 185 FPL",
    "% food insecure children in HH w/ HH incomes above 185 FPL"
]

X = ga_2023[features].dropna()

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Results:", pca.explained_variance_ratio_)
```

```python
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])

plt.figure(figsize=(8, 6))
plt.scatter(pca_df["PC1"], pca_df["PC2"], alpha=0.6)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of Child Food Insecurity Factors")

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12, 5))

# Child Food Insecurity
plt.subplot(1, 2, 1)
sns.histplot(ga_2023["Child Food Insecurity Rate"], kde=True)
plt.title("Distribution of Child Food Insecurity Rate")

# Poverty %
plt.subplot(1, 2, 2)
sns.histplot(ga_2023["% food insecure children in HH w/ HH incomes below 185 FPL"], kde=True)
plt.title("Distribution of Children Below 185% FPL")

plt.tight_layout()
plt.show()
```
